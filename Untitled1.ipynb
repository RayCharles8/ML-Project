```python

class Config:
    csv_path = "/content/drive/MyDrive/Project/Dataset.csv"
    cif_root = "/content/drive/MyDrive/Project/cif_files"

    cutoff = 8.0          # neighbor cutoff (Å)
    max_neighbors = 12

    gaussian_dim = 50     # edge feature expansion
    sigma = 0.5

    batch_size = 32
    lr = 1e-3
    epochs = 50

    device = "cuda" if torch.cuda.is_available() else "cpu"

cfg = Config()
print("Using device:", cfg.device)
```

```python
class GaussianDistance:
    def __init__(self, dmin=0, dmax=8, step=0.2, sigma=0.5):
        self.filter_centers = np.arange(dmin, dmax + step, step)
        self.sigma = sigma

    def expand(self, distances):
        distances = np.array(distances)
        return np.exp(-(distances[..., None] - self.filter_centers)**2 / self.sigma**2)

gaussian_expander = GaussianDistance(
    dmin=0,
    dmax=cfg.cutoff,
    step=cfg.cutoff / cfg.gaussian_dim,
    sigma=cfg.sigma
)
```

```python
def get_atom_features(element):
    el = Element(element)

    return [
        el.Z,                                # atomic number
        el.X if el.X else 0,                 # electronegativity
        el.atomic_radius if el.atomic_radius else 0,
        el.group if el.group else 0,
        el.row if el.row else 0              # period
    ]
```

```python
class CrystalDataset(Dataset):
    def __init__(self, csv_file, root_dir):
        super().__init__()
        self.df = pd.read_csv(csv_file)
        self.root_dir = root_dir

    def len(self):
        return len(self.df)

    def get(self, idx):
        row = self.df.iloc[idx]

        cif_path = row["cif"]
        structure = Structure.from_file(cif_path)

        # =========================
        # Node features
        # =========================
        atom_features = []
        for site in structure:
            atom_features.append(get_atom_features(site.specie.symbol))

        x = torch.tensor(atom_features, dtype=torch.float)

        # =========================
        # Edge construction
        # =========================
        edge_index = []
        edge_attr = []

        for i, site in enumerate(structure):
            neighbors = structure.get_neighbors(site, cfg.cutoff)

            neighbors = sorted(neighbors, key=lambda x: x[1])[:cfg.max_neighbors]

            for neighbor in neighbors:
                j = neighbor.index
                dist = neighbor.nn_distance

                edge_index.append([i, j])

                # Gaussian expansion
                edge_feature = gaussian_expander.expand([dist])[0]
                edge_attr.append(edge_feature)

        edge_index = torch.tensor(edge_index, dtype=torch.long).t().contiguous()
        edge_attr = torch.tensor(edge_attr, dtype=torch.float)

        # =========================
        # Label
        # =========================
        y = torch.tensor([row["label"]], dtype=torch.float)

        return Data(x=x, edge_index=edge_index, edge_attr=edge_attr, y=y)
```

```python
class CGCNNConv(MessagePassing):
    def __init__(self, node_dim, edge_dim):
        super().__init__(aggr='add')

        self.fc = nn.Linear(2 * node_dim + edge_dim, node_dim)
        self.bn = nn.BatchNorm1d(node_dim)

    def forward(self, x, edge_index, edge_attr):
        return self.propagate(edge_index, x=x, edge_attr=edge_attr)

    def message(self, x_i, x_j, edge_attr):
        z = torch.cat([x_i, x_j, edge_attr], dim=1)
        return F.softplus(self.fc(z))

    def update(self, aggr_out, x):
        return self.bn(x + aggr_out)
```

```python
class CGCNN(nn.Module):
    def __init__(self, node_dim=5, edge_dim=50, hidden_dim=64):
        super().__init__()

        self.embedding = nn.Linear(node_dim, hidden_dim)

        self.conv1 = CGCNNConv(hidden_dim, edge_dim)
        self.conv2 = CGCNNConv(hidden_dim, edge_dim)
        self.conv3 = CGCNNConv(hidden_dim, edge_dim)

        self.pool = global_mean_pool

        self.fc1 = nn.Linear(hidden_dim, 64)
        self.fc2 = nn.Linear(64, 1)

    def forward(self, data):
        x, edge_index, edge_attr, batch = \
            data.x, data.edge_index, data.edge_attr, data.batch

        x = F.relu(self.embedding(x))

        x = self.conv1(x, edge_index, edge_attr)
        x = self.conv2(x, edge_index, edge_attr)
        x = self.conv3(x, edge_index, edge_attr)

        x = self.pool(x, batch)

        x = F.relu(self.fc1(x))
        x = self.fc2(x)

        return x.view(-1)
```

This implementation follows:

Key design:
- Node features → atomic information

- Global features → dataset descriptors (bandgap, lattice, etc.)

Why MEGNet:
- Supports global state vector (critical for this dataset)
- Proven high accuracy on Materials Project benchmarks

```python
import megnet.callbacks as megnet_callbacks
import numpy as np
import tensorflow.keras.callbacks as kc

# Reimplement the __init__ for ModelCheckpointMAE to use np.inf
def patched_mcmae_init(self, filepath, monitor='val_mae', verbose=0,
                     save_best_only=False, save_weights_only=False,
                     val_gen=None, steps_per_val=None, target_scaler=None,
                     period=1, mode='min'):

    super(type(self), self).__init__() # Call the parent Callback's init

    self.monitor = monitor
    self.verbose = verbose
    self.filepath = filepath
    self.save_best_only = save_best_only
    self.save_weights_only = save_weights_only
    self.period = period
    self.epochs_since_last_save = 0
    self.val_gen = val_gen
    self.steps_per_val = steps_per_val
    self.target_scaler = target_scaler

    if mode == 'min':
        self.monitor_op = np.less
        self.best = np.inf  # Corrected to np.inf
    elif mode == 'max':
        self.monitor_op = np.greater
        self.best = -np.inf # Corrected to -np.inf
    else:
        if 'acc' in self.monitor or self.monitor.startswith('fmeasure'):
            self.monitor_op = np.greater
            self.best = -np.inf # Corrected to -np.inf
        else:
            self.monitor_op = np.less
            self.best = np.inf  # Corrected to np.inf

# Apply the patch by replacing the original __init__ method
megnet_callbacks.ModelCheckpointMAE.__init__ = patched_mcmae_init

print("✅ ModelCheckpointMAE __init__ patched successfully for NumPy 2.0 compatibility")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch_geometric.data import Data, Dataset
from torch_geometric.loader import DataLoader
from torch_geometric.nn import MessagePassing, global_mean_pool

from pymatgen.core import Structure
import numpy as np
from tqdm import tqdm
import os

# Atomic feature function (based on CGCNN literature)
def get_atom_features(site):
    el = site.specie

    return [
        el.Z,                       # atomic number
        el.X if el.X else 0,        # electronegativity
        el.atomic_mass,
        el.atomic_radius if el.atomic_radius else 0
    ]

# Build graph from CIF
def build_graph(cif_path, label, cutoff=4.0):
    try:
        structure = Structure.from_file(cif_path)

        node_features = []
        edge_index = []
        edge_attr = []

        # Node features
        for site in structure:
            node_features.append(get_atom_features(site))

        # Build edges
        for i, site_i in enumerate(structure):
            neighbors = structure.get_neighbors(site_i, cutoff)

            for neighbor in neighbors:
                j = neighbor.index
                dist = neighbor.nn_distance

                edge_index.append([i, j])
                edge_attr.append([dist])

        x = torch.tensor(node_features, dtype=torch.float)
        edge_index = torch.tensor(edge_index, dtype=torch.long).t().contiguous()
        edge_attr = torch.tensor(edge_attr, dtype=torch.float)
        y = torch.tensor([label], dtype=torch.float)

        return Data(x=x, edge_index=edge_index, edge_attr=edge_attr, y=y)

    except Exception as e:
        return None

base_path = "/content/drive/MyDrive/Project/"

data_list = []

for i, row in tqdm(df.iterrows(), total=len(df)):
    graph = build_graph(
        os.path.join(base_path, row['cif']),
        row['label']
    )

    if graph is not None:
        data_list.append(graph)

print("Total graphs:", len(data_list))
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    data_list,
    test_size=0.2,
    random_state=42
)

train_loader = DataLoader(train_data, batch_size=32, shuffle=True)
test_loader = DataLoader(test_data, batch_size=32)
```

```python
class CGCNNLayer(MessagePassing):
    def __init__(self, in_channels, out_channels):
        super().__init__(aggr='mean')  # standard in CGCNN

        self.lin_node = nn.Linear(in_channels, out_channels)
        self.lin_edge = nn.Linear(1, out_channels)

    def forward(self, x, edge_index, edge_attr):
        return self.propagate(edge_index, x=x, edge_attr=edge_attr)

    def message(self, x_j, edge_attr):
        edge_feat = self.lin_edge(edge_attr)
        return x_j + edge_feat

    def update(self, aggr_out):
        return F.relu(self.lin_node(aggr_out))
```

```python
class CharlesCGCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.conv1 = CGCNNLayer(4, 64)
        self.conv2 = CGCNNLayer(64, 64)

        self.fc1 = nn.Linear(64, 32)
        self.fc2 = nn.Linear(32, 1)

    def forward(self, data):
        x, edge_index, edge_attr, batch = \
            data.x, data.edge_index, data.edge_attr, data.batch

        x = self.conv1(x, edge_index, edge_attr)
        x = self.conv2(x, edge_index, edge_attr)

        x = global_mean_pool(x, batch)

        x = F.relu(self.fc1(x))
        x = torch.sigmoid(self.fc2(x))

        return x
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

model = CharlesCGCNN().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.BCELoss()

for epoch in range(50):
    model.train()
    total_loss = 0

    for batch in train_loader:
        batch = batch.to(device)

        optimizer.zero_grad()
        out = model(batch).view(-1)
        loss = criterion(out, batch.y)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss:.4f}")
```

```python
from sklearn.metrics import accuracy_score, f1_score, precision_score, roc_auc_score

model.eval()

y_true = []
y_pred = []
y_prob = []

with torch.no_grad():
    for batch in test_loader:
        batch = batch.to(device)
        out = model(batch).view(-1)

        y_true.extend(batch.y.cpu().numpy())
        y_prob.extend(out.cpu().numpy())
        y_pred.extend((out > 0.5).cpu().numpy())

print("Accuracy:", accuracy_score(y_true, y_pred))
print("F1:", f1_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred))
print("ROC-AUC:", roc_auc_score(y_true, y_prob))
```

# The Model

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import MessagePassing, global_mean_pool

class MEGNetLayer(MessagePassing):
    def __init__(self, node_dim, edge_dim, hidden_dim):
        super().__init__(aggr='add')  # sum aggregation
        self.edge_mlp = nn.Sequential(
            nn.Linear(2*node_dim + edge_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, edge_dim),
            nn.ReLU()
        )
        self.node_mlp = nn.Sequential(
            nn.Linear(node_dim + edge_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, node_dim),
            nn.ReLU()
        )

    def forward(self, x, edge_index, edge_attr):
        # x: node features [N, node_dim]
        # edge_index: [2, E]
        # edge_attr: [E, edge_dim]
        row, col = edge_index
        # concat sender + receiver + edge_attr
        edge_inputs = torch.cat([x[row], x[col], edge_attr], dim=1)
        edge_attr = self.edge_mlp(edge_inputs)

        # aggregate messages
        aggr_messages = self.propagate(edge_index, x=x, edge_attr=edge_attr)
        x = self.node_mlp(torch.cat([x, aggr_messages], dim=1))
        return x, edge_attr

    def message(self, x_j, edge_attr):
        return edge_attr  # message is edge feature
```

```python
class MEGNetModel(nn.Module):
    def __init__(self, node_dim=1, edge_dim=1, hidden_dim=64, num_layers=3, global_features_dim=0):
        super().__init__()

        self.layers = nn.ModuleList()
        for _ in range(num_layers):
            self.layers.append(MEGNetLayer(node_dim, edge_dim, hidden_dim))

        self.pool = global_mean_pool  # simple mean pooling

        self.fc = nn.Sequential(
            nn.Linear(node_dim + global_features_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 2)  # binary classification: Topological / Trivial
        )

    def forward(self, data):
        x, edge_index, edge_attr, batch = data.x, data.edge_index, data.edge_attr, data.batch
        for layer in self.layers:
            x, edge_attr = layer(x, edge_index, edge_attr)

        # global pooling
        x = self.pool(x, batch)

        # add global features if any
        if hasattr(data, "global_features"):
            x = torch.cat([x, data.global_features], dim=1)

        out = self.fc(x)
        return out
```

```python
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, classification_report
import torch.optim as optim

def train_epoch(model, loader, criterion, optimizer, device):
    model.train()
    losses = []
    all_preds, all_labels = [], []
    for data in loader:
        data = data.to(device)
        optimizer.zero_grad()
        out = model(data)
        loss = criterion(out, data.y.view(-1))
        loss.backward()
        optimizer.step()
        losses.append(loss.item())

        preds = torch.argmax(out, dim=1).detach().cpu()
        all_preds.extend(preds.numpy())
        all_labels.extend(data.y.view(-1).cpu().numpy())
    return sum(losses)/len(losses), all_preds, all_labels

def evaluate(model, loader, criterion, device):
    model.eval()
    losses = []
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for data in loader:
            data = data.to(device)
            out = model(data)
            loss = criterion(out, data.y.view(-1))
            losses.append(loss.item())

            probs = torch.softmax(out, dim=1)[:,1].cpu().numpy()
            preds = torch.argmax(out, dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(data.y.view(-1).cpu().numpy())
            all_probs.extend(probs)
    acc = accuracy_score(all_labels, all_preds)
    f1 = f1_score(all_labels, all_preds)
    roc = roc_auc_score(all_labels, all_probs)
    report = classification_report(all_labels, all_preds)
    return sum(losses)/len(losses), acc, f1, roc, report
```

```python
# Train and test the model on the sample data
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = MEGNetModel().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)

num_epochs = 30

for epoch in range(num_epochs):
    train_loss, train_pred, _train_label = train_epoch(model, sample_train_loader, criterion, optimizer, device)
    val_loss, val_acc, val_f1, val_roc, val_report = evaluate(model, sample_val_loader, criterion, device)
    print(f"Epoch {epoch+1}: Train Loss={train_loss:.4f}, Val Loss={val_loss:.4f}, Acc={val_acc:.3f}, F1={val_f1:.3f}, ROC-AUC={val_roc:.3f}")
```
