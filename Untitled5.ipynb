```python
import os
import numpy as np
import pandas as pd

from pymatgen.core import Structure
from pymatgen.io.cif import CifParser

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch_geometric.data import Data, Dataset # Added Dataset import
from torch_geometric.loader import DataLoader
from torch_geometric.nn import CGConv, global_mean_pool
from torch_geometric.nn.pool import radius_graph

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, classification_report

import warnings
warnings.filterwarnings("ignore")
```

Data pre-processing

```python
# Load CSV
df = pd.read_csv(DATA_PATH)
print("Original columns:", df.columns.tolist())

# Ensure 'cif' column is always a string, handling potential NaN values
df['cif'] = df['cif'].fillna('').astype(str) # Fill NaN with empty string and convert to str

# Encode categorical columns as integer codes, excluding 'cif'
# because 'cif' contains file paths which should remain strings.
object_cols = df.select_dtypes(include=['object']).columns
for col in object_cols:
    if col != 'cif':  # Exclude 'cif' column from categorical encoding
        df[col] = df[col].astype('category').cat.codes

# Show a few entries and class balance
print("\nSample data:")
print("\nClass distribution:", df['label'].value_counts().to_dict())
```

```python
# Stratified split: 70% train, 15% val, 15% test
train_df, temp_df = train_test_split(
    df, test_size=0.30, random_state=42, stratify=df['label']
)
val_df, test_df = train_test_split(
    temp_df, test_size=0.50, random_state=42, stratify=temp_df['label']
)

print(f"Train: {len(train_df)}  Val: {len(val_df)}  Test: {len(test_df)}")
print("Label distribution (train/val/test):",
      train_df['label'].mean(), val_df['label'].mean(), test_df['label'].mean())
```

Build Crysal Graphs

```python
def get_global_features(row):
    """Extract global features (12 dims) from a DataFrame row."""
    cols = ['spacegroup_number','spacegroup','number of atoms','Band Gap',
            'a','b','c','alpha','beta','gamma','Z','electronegativity']
    vals = [row[col] for col in cols]
    return torch.tensor(vals, dtype=torch.float).unsqueeze(0)  # shape [1, 11]

def build_graph(row):
    """Create a torch_geometric Data object from one row of the DataFrame."""
    # Parse the CIF structure
    cif_files = CIF_DIR + row['cif'].split('/')[-1]
    structure = CifParser(cif_files).get_structures()[0]
    # Node features: atomic number Z and electronegativity
    atom_Z = [site.specie.Z for site in structure]
    atom_X = [site.specie.X if site.specie.X is not None else 0.0 for site in structure]
    x = torch.tensor(list(zip(atom_Z, atom_X)), dtype=torch.float)
    # Edge index by spatial radius graph
    pos = torch.tensor(structure.cart_coords, dtype=torch.float)
    edge_index = radius_graph(pos, r=5.0, loop=False)
    src, dst = edge_index
    edge_attr = torch.norm(pos[src] - pos[dst], dim=1, keepdim=True)
    # Global features and label
    u = get_global_features(row)
    y = torch.tensor([row['label']], dtype=torch.long)
    return Data(x=x, edge_index=edge_index, edge_attr=edge_attr, u=u, y=y)

# Test on first training sample
sample_data = build_graph(train_df.iloc[0])
print(sample_data)
print("x:", sample_data.x.shape, "edge_index:", sample_data.edge_index.shape,
      "edge_attr:", sample_data.edge_attr.shape, "u:", sample_data.u.shape, "y:", sample_data.y)
```

```python
# Create Dataset classes for train/val/test
class CrystalDataset(Dataset):
    def __init__(self, df):
        super().__init__()
        self.df = df.reset_index(drop=True)
    def len(self):
        return len(self.df)
    def get(self, idx):
        return build_graph(self.df.iloc[idx])

train_dataset = CrystalDataset(train_df)
val_dataset   = CrystalDataset(val_df)
test_dataset  = CrystalDataset(test_df)

print("Datasets sizes:", len(train_dataset), len(val_dataset), len(test_dataset))
```

```python
import torch
import torch.nn as nn
from torch_geometric.nn import global_mean_pool, CGConv


class CharlesModel(nn.Module):

    def __init__(self, node_dim=2, edge_dim=1, global_dim=11):
        super().__init__()

        hidden = 64

        # project node features
        self.node_embedding = nn.Linear(node_dim, hidden)

        # CGCNN convolution layers
        self.conv1 = CGConv(hidden, edge_dim)
        self.conv2 = CGConv(hidden, edge_dim)
        self.conv3 = CGConv(hidden, edge_dim)

        # global feature embedding
        self.global_fc = nn.Linear(global_dim, 32)

        # classifier
        self.fc1 = nn.Linear(hidden + 32, 64)
        self.fc2 = nn.Linear(64, 2)

        self.relu = nn.ReLU()

    def forward(self, data):

        x = data.x
        edge_index = data.edge_index
        edge_attr = data.edge_attr
        batch = data.batch
        u = data.u

        # node embedding
        x = self.relu(self.node_embedding(x))

        # graph convolutions
        x = self.relu(self.conv1(x, edge_index, edge_attr))
        x = self.relu(self.conv2(x, edge_index, edge_attr))
        x = self.relu(self.conv3(x, edge_index, edge_attr))

        # pooling
        x = global_mean_pool(x, batch)

        # global features
        u = self.relu(self.global_fc(u))

        # concatenate
        x = torch.cat([x, u], dim=1)

        # classifier
        x = self.relu(self.fc1(x))
        x = self.fc2(x)

        return x
```

```python
class CharlesCGCNN(nn.Module):
    def __init__(self, hidden_dim=64):
        super().__init__()
        self.conv1 = CGConv(channels=(2, hidden_dim), dim=1)
        self.conv2 = CGConv(channels=(hidden_dim, hidden_dim), dim=1)
        self.fc = nn.Linear(hidden_dim + 12, 2) # Corrected from hidden_dim + 11 to hidden_dim + 12

    def forward(self, data):
        x, edge_index, edge_attr, batch, u = (
            data.x, data.edge_index, data.edge_attr, data.batch, data.u
        )
        # Two CGConv layers + ReLU
        x = F.relu(self.conv1(x, edge_index, edge_attr))
        x = F.relu(self.conv2(x, edge_index, edge_attr))
        # Global mean pooling
        x = global_mean_pool(x, batch)
        # Concatenate global features and classify
        u = u.squeeze(1)
        out = torch.cat([x, u], dim=1)
        return self.fc(out)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = CharlesCGCNN(hidden_dim=64).to(device)
print(model)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from pymatgen.io.cif import CifParser
import torch
from torch_geometric.data import Data

# Load dataset (assumes Dataset.csv is in current directory or Google Drive)
df = pd.read_csv("/content/drive/MyDrive/Project/Dataset.csv")

# Encode categorical spacegroup (if needed)
df['spacegroup_cat'] = df['spacegroup'].astype('category').cat.codes

# Stratified split: train/validation/test
train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['label'], random_state=42)
valid_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['label'], random_state=42)

def make_graph_entry(row):
    """Parse CIF and construct PyG Data for one material."""
    parser = CifParser(row['cif'])
    struct = parser.get_structures()[0]
    # Extract node features: [atomic number, electronegativity] per atom
    atom_feats = []
    for site in struct:
        Z = site.specie.Z
        # If X (electronegativity) attribute is not present, we can get it from periodic table
        # Pymatgen Element has .X (Pauling), but just using specie if available:
        X = site.specie.X if site.specie.X is not None else 0.0
        atom_feats.append([float(Z), float(X)])
    x = torch.tensor(atom_feats, dtype=torch.float)
    # Build edges: connect neighbors within cutoff (e.g. 5 Å)
    edge_index = []
    edge_attr = []
    cutoff = 5.0
    for i, site in enumerate(struct):
        neighbors = struct.get_neighbors(site, cutoff)
        for neigh in neighbors:
            j = neigh.index
            dist = neigh.nn_distance
            # add edge i->j with feature = distance
            edge_index.append([i, j])
            edge_attr.append([dist])
    # Convert to tensors
    edge_index = torch.tensor(edge_index, dtype=torch.long).t().contiguous()
    edge_attr = torch.tensor(edge_attr, dtype=torch.float)
    # Global feature vector u (11 dims as specified)
    u = torch.tensor([
        float(row['spacegroup_number']),
        float(row['number of atoms']),
        float(row['Band Gap']),
        float(row['a']), float(row['b']), float(row['c']),
        float(row['alpha']), float(row['beta']), float(row['gamma']),
        float(row['Z']), float(row['electronegativity'])
    ], dtype=torch.float)
    y = torch.tensor([int(row['label'])], dtype=torch.long)
    return Data(x=x, edge_index=edge_index, edge_attr=edge_attr, u=u, y=y)

# Create Data lists for each split
train_graphs = [make_graph_entry(row) for _, row in train_df.iterrows()]
valid_graphs = [make_graph_entry(row) for _, row in valid_df.iterrows()]
test_graphs  = [make_graph_entry(row) for _, row in test_df.iterrows()]
```
